# Ollama Embeddings

```python
import ollama

# Pull the model first if you haven't (nomic-embed-text is highly recommended)
# ollama.pull('nomic-embed-text')

response = ollama.embed(
    model='qwen3-embedding:4b',
    input='The sky is blue because of Rayleigh scattering'
)

# Access the embedding vector
embedding = response.embeddings[0]
print(f"Vector Length: {len(embedding)}")
print(f"First 5 values: {embedding[:5]}")
```

# Ollama Generation

```python
response = ollama.generate(model='qwen3:8b', prompt='Why is the sky blue?')
print(response['response'])
```

# OpenAI API

```python
from openai import OpenAI

client = OpenAI(
    base_url='http://127.0.0.1:11434/v1',
    api_key = "ollama",
)

response = client.chat.completions.create(
    model='qwen3:8b',
    messages=[
        {"role": "system", "content": "Ты промпт-инженер"},
        {"role": "user", "content": "Составь шаблон промпта, используя цепочку размышлений"}
    ]
)
print(response.choices[0].message.content)
```
