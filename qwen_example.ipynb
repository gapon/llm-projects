```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

API_KEY = os.getenv("CD_OLLAMA_API_KEY")
```

```python
client = OpenAI(
    base_url='https://llm.citydrive.tech/api',
    api_key=API_KEY,
)

response = client.chat.completions.create(
    model='qwen3.6:27b',
    messages=[
        {"role": "system", "content": "Ты промпт-инженер"},
        {"role": "user", "content": "Составь шаблон промпта, используя цепочку размышлений"}
    ]
)
print(response.choices[0].message.content)
```
